```python
import pandas as pd
import matplotlib.pyplot as plt

# Đọc tệp CSV (thay 'your_file.csv' bằng tên tệp của bạn)
file_path = 'orders_with_1_items.csv'  # Đường dẫn đến tệp của bạn
data = pd.read_csv(file_path)

# Kiểm tra số lượng hàng
num_rows = data.shape[0]
print(f"Số lượng hàng trong tệp: {num_rows}")


# Kiểm tra và tính toán trung bình, trung vị
if 'KPT duration (minutes)' in data.columns:
    kpt_column = data['KPT duration (minutes)'].dropna()
    mean_kpt = kpt_column.mean()
    median_kpt = kpt_column.median()
    print(f"Trung bình KPT duration (minutes): {mean_kpt}")
    print(f"Trung vị KPT duration (minutes): {median_kpt}")
    
    # Vẽ biểu đồ histogram
    plt.hist(kpt_column, bins=20, color='blue', edgecolor='black')
    plt.title('Histogram of KPT duration (minutes)')
    plt.xlabel('KPT duration (minutes)')
    plt.ylabel('Frequency')
    plt.axvline(mean_kpt, color='red', linestyle='dashed', linewidth=1, label=f'Mean: {mean_kpt:.2f}')
    plt.axvline(median_kpt, color='green', linestyle='dashed', linewidth=1, label=f'Median: {median_kpt:.2f}')
    plt.legend()
    plt.show()
else:
    print("Cột 'KPT duration (minutes)' không tồn tại trong tệp.")
```
