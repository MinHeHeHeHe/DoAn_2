```python
import pandas as pd

# Đọc tệp CSV
df = pd.read_csv('order_history_kaggle_data.csv')

# Lọc các hàng có Order Ready Marked = Correctly
filtered_df = df[df['Order Ready Marked'] == 'Correctly']

# Kiểm tra xem cột "items in order" có tồn tại không
if 'Items in order' in filtered_df.columns:
    print("Cột 'items in order' tồn tại!")
    # In nội dung của cột
    print(filtered_df['Items in order'])
else:
    print("Cột 'items in order' không tồn tại.")
```

```python
# Tính tổng số lượng món ăn trong từng đơn hàng
filtered_df['Order Item Count'] = filtered_df['Items in order'].str.findall(r'(\d+)\s*x\s*[^,]+').apply(lambda x: sum(map(int, x)) if x else 0)

# Hiển thị kết quả
print(filtered_df[['Items in order', 'Order Item Count']])
```

```python
# Tạo thư mục để lưu các file nhỏ (nếu cần)
import os
output_dir = "split_orders_correctly"
os.makedirs(output_dir, exist_ok=True)

# Lấy danh sách các giá trị duy nhất của số lượng món ăn
unique_item_counts = filtered_df['Order Item Count'].unique()

# Lặp qua từng số lượng món ăn và lưu các file nhỏ
for count in unique_item_counts:
    subset = filtered_df[filtered_df['Order Item Count'] == count]
    output_file = os.path.join(output_dir, f"orders_with_{count}_items.csv")
    subset.to_csv(output_file, index=False)
    print(f"Đã lưu file: {output_file}")
```
